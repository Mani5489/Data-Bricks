```python
from pyspark.sql import functions as f, Window
from pyspark.sql.types import *

data = [(1, 1), (2, 1), (3, 1), (4, 2), (5, 1), (6, 2), (7, 2)]

schema = StructType([
    StructField('id', IntegerType()),
    StructField('num', IntegerType())
])

logs = spark.createDataFrame(data=data,schema=schema)

w1 = Window.orderBy('id')
w2 = Window.partitionBy('num').orderBy('id')

res = logs.withColumn('gap',(f.row_number().over(w1)-f.row_number().over(w2)))
res = res.groupBy('num','gap').agg(f.count('*').alias('cnt'))
res = res.filter(f.col('cnt')>=3).select('num')
res.show()

```
