```python
from pyspark.sql import functions as f
from pyspark.sql.types import *

data = [(101, 'Rahul', 'rahul@gmail.com', 'Bangalore', 50000), (102, 'Amit', None, 'Pune', 60000), (103, None, 'john@gmail.com', 'Chennai', 70000), (104, 'David', 'david@gmail.com', None, None), (105, None, None, None, 80000), (None, 'Ravi', 'ravi@gmail.com', 'Hyderabad', 55000)]

schema = StructType(
    [
        StructField('customer_id',IntegerType()),
        StructField('name',StringType()),
        StructField('email',StringType()),
        StructField('city',StringType()),
        StructField('salary',IntegerType()),
    ]
)

customers = spark.createDataFrame(data=data,schema=schema)
customers = customers.dropna(subset='customer_id')
customers = customers.fillna(0,subset='salary')
customers = customers.fillna('unknown',subset='name')
customers = customers.fillna("Not Available",subset='email')
customers = customers.fillna("Unknown City",subset='city')
customers = customers.withColumn('data_quality_status',f.lit('PASS'))


customers.write.format('delta').mode('overwrite').saveAsTable('customers')



```
