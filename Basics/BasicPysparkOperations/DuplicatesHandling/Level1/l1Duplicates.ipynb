```python
from pyspark.sql import functions as f
from pyspark.sql.types import *

data = [
    (101, 'Rahul', 'Bangalore'),
    (101, 'Rahul', 'Bangalore'),
    (102, 'Amit', 'Pune'),
    (103, 'John', 'Chennai'),
    (103, 'John', 'Chennai'),
    (103, 'John', 'Bangalore'),
    (103, 'John', 'Bangalore')
]

schema = StructType([
    StructField('customer_id',IntegerType(),False),
    StructField('name',StringType(),True),
    StructField('city',StringType(),True)
])

customers = spark.createDataFrame(data=data,schema=schema)
print(f'before:{customers.select(f.col('customer_id')).count()}')
customers = customers.dropDuplicates(['customer_id','city']) #Note gives 4 records
# customers = customers.dropDuplicates() gives 3 records
print(f'after:{customers.select(f.col('customer_id')).count()}')
customers.show()


```
