```python
customers = spark.read.table('madstagescd2.madstage.silver_customers')
products = spark.read.table('madstagescd2.madstage.silver_products')
orders = spark.read.table('madstagescd2.madstage.silver_orders')

# order_id|customer_id|product_id|quantity|order_amount|order_date|   status|is_current|updated_dt|created_dt|
#|product_id|       product_name|   category|    brand|price|status|is_current|updated_dt|created_dt
#customer_id|customer_name|     city|customer_segment|           email|     phone|is_current|updated_dt|created_dt

gold = orders.alias('o').join(customers.alias('c'),on='customer_id',how='left').join(products.alias('p'),on='product_id',how='left').select('o.*','p.product_name','p.category','p.brand','p.price','c.customer_name','c.city','c.customer_segment','c.email','c.phone')

gold.write.format('delta').mode('overwrite').option('overwriteSchema','true').saveAsTable('madstagescd2.madstage.goldCustomersOrders')

print('Completed')
```
