```python
from pyspark.sql import functions as f
from pyspark.sql.types import *
import re
from datetime import datetime

spr = spark.read.table('madstagescd2.madstage.goldcustomersorders')

spr = spr.groupBy('customer_id','customer_name').agg(
    f.sum(f.col('order_amount')).alias('total_order_amount'),
    f.count(f.col('order_id')).alias('total_order_id'),
    f.sum(f.col('quantity')).alias('total_quantity'),
    f.avg(f.col('order_amount')).alias('avg_order_amount'),
)

spr.show()

spr.write.format('delta').mode('overwrite').option('overwriteSchema','true').saveAsTable('SalesPerformanceReport')

print('SalesPerformanceReport Generated')
```
