```python
from pyspark.sql import functions as f,Window
import re
from datetime import datetime
from pyspark.sql.types import *

c_schema = StructType([
    StructField('customer_id',IntegerType()),
    StructField('customer_name',StringType()),
    StructField('city',ArrayType(StringType()))
])

data1 = [
    (1,'John',['Banglore','Bellary']),
    (2,'Sanjay',['Banglore','Bellary'])
]

customer = spark.createDataFrame(data=data1,schema=c_schema)
customer_new = customer.withColumn('city',f.explode('city'))
customer_new.show()
```
