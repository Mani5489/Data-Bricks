```python
from pyspark.sql import functions as f
from pyspark.sql.types import *
import re
from datetime import datetime

data = [
    (1,1,'Match','M','K','MisMatch',26,26,'MisMatch',400,500,'MisMatch'),
    (1,1,'Match','M','K','MisMatch',26,26,'MisMatch',400,500,'MisMatch'),
    (1,1,'Match','M','K','MisMatch',26,26,'MisMatch',400,500,'MisMatch')
]

schema = StructType(
    [
        StructField('id_src',IntegerType()),
        StructField('id_tgt',IntegerType()),
        StructField('id_match',StringType()),
        StructField('name_src',StringType()),
        StructField('name_tgt',StringType()),
        StructField('name_match',StringType()),
        StructField('age_src',IntegerType()),
        StructField('age_tgt',IntegerType()),
        StructField('age_match',StringType()),
        StructField('salary_src',IntegerType()),
        StructField('salary_tgt',IntegerType()),
        StructField('salary_match',StringType()),
    ]
)

df = spark.createDataFrame(data=data,schema=schema)

c = [col for col in df.columns if col.endswith('_match')]

# find full mismatch columns
mismatched = []


for i in c:
    val = df.filter(f.col(i) == 'MisMatch').count()
    if val > 0:
        mismatched.append(i)

print(f"mimatched columns {mismatched}")
```
